```python
import pandas as pd
df = pd.read_csv("/content/oil_by_country.csv")
print(df.isnull())
```

```python


print(f"DataFrame Columns: {df.columns.tolist()}")


column_to_analyze = 'Production '


Q1 = df[column_to_analyze].quantile(0.25)
Q3 = df[column_to_analyze].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df[column_to_analyze] < lower) |
              (df[column_to_analyze] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

num_cols = df.select_dtypes(include=['int64', 'float64']).columns

df[num_cols] = scaler.fit_transform(df[num_cols])

print(df[num_cols])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

# The correct column name for the country column is 'Country' (without a trailing space).
# Other columns like 'Production ', 'Consumption ', 'Exports ', 'Imports ' do have trailing spaces.
df['Country'] = encoder.fit_transform(df['Country'])

print(df)
```
