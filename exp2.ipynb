```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("/content/Vrinda Store Data Analysis.xlsx")

category_count = df['Category'].value_counts()

plt.figure(figsize=(10,6))
plt.bar(category_count.index, category_count.values)
plt.title("Distribution of Product Categories in Vrinda Store")
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Using the oil_by_country.csv file available in the Colab environment
df_oil = pd.read_csv("/content/oil_by_country.csv")

# Plotting the distribution of 'Country' as 'type' column does not exist in this dataset
country_count = df_oil['Country'].value_counts().head(20) # Showing top 20 countries for better visualization

plt.figure(figsize=(12, 6))
plt.bar(country_count.index, country_count.values)
plt.title("Distribution of Oil Producing Countries (Top 20)")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=90, ha='right') # Rotate labels for readability
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Group by country and sum 'Proven_reserves', then get the top 10
top_10_reserves = df_oil.groupby('Country')['Proven_reserves'].sum().nlargest(10)

plt.figure(figsize=(12, 7))
top_10_reserves.plot(kind='bar', color='skyblue')
plt.title('Top 10 Countries by Proven Oil Reserves')
plt.xlabel('Country')
plt.ylabel('Proven Reserves (in billions)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
